```python
# Step 1: Install TeX Live LaTeX Compiler
!apt-get update -qq
!apt-get install -y -qq texlive-latex-base texlive-latex-extra texlive-fonts-recommended texlive-publishers > /dev/null
print("TeX Live installed successfully!")
```

```python
# Step 2: Upload Report Zip Packages
from google.colab import files
import os, zipfile, subprocess

print("Please upload the 3 report packages:")
print("1. 70177784_A2_Report_Package.zip")
print("2. 70177784_B3_Report_Package.zip")
print("3. 70177784_C3_Report_Package.zip")
uploaded = files.upload()
```

```python
# Step 3: Extract and Compile PDFs
sap_id = '70177784'
tasks = ['A2', 'B3', 'C3']

for t in tasks:
    zip_name = f"{sap_id}_{t}_Report_Package.zip"
    folder = f"build_{t}"
    os.makedirs(folder, exist_ok=True)
    if zip_name in uploaded:
        with zipfile.ZipFile(zip_name, 'r') as zf:
            zf.extractall(folder)
        print(f"Compiling {sap_id}_{t}_Report.pdf...")
        cmd = f"cd {folder} && pdflatex -interaction=nonstopmode {sap_id}_{t}_Report.tex && pdflatex -interaction=nonstopmode {sap_id}_{t}_Report.tex"
        subprocess.run(cmd, shell=True, check=True)
        pdf_out = f"{folder}/{sap_id}_{t}_Report.pdf"
        if os.path.exists(pdf_out):
            print(f"[✓] Compiled successfully: {pdf_out}")
            files.download(pdf_out)
        else:
            print(f"[!] Failed to generate {pdf_out}")
```
